# 01 — EDA LegitPhish

Carga el CSV, invierte la etiqueta (1 = phishing, 0 = legítima) y resume las 16 features.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "ml").exists():
    ROOT = ROOT.parent.parent if (ROOT.parent.parent / "ml").exists() else ROOT
sys.path.insert(0, str(ROOT))

from ml.src.data.load_data import load_dataset

df = load_dataset(ROOT / "ml" / "data" / "raw" / "legitphish.csv")
df.head()
```

```python
print(df.shape)
print(df["label"].value_counts().rename({1: "phishing", 0: "legitima"}))
df.describe().T
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(figsize=(6, 4))
sns.countplot(x=df["label"].map({1: "phishing", 0: "legitima"}), ax=ax)
ax.set_title("Distribución tras invertir ClassLabel")
plt.show()
```
